```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn  as sns
```

```python
df = pd.read_csv('/content/healthcare_dataset.csv')
df
```

```python
print("\nDataset Information:")
print(df.info())
```

```python
print("\nMissing Values:")
print(df.isnull().sum())

```

```python
df['Medical_Code'].isnull()
```

```python
df['Standardized_Medical_Code'] = df['Medical_Code'].str.replace('ICD-10-', '')
print(df[['Medical_Code', 'Standardized_Medical_Code']].head())
```

```python
df['Admission_Date'] = pd.to_datetime(df['Admission_Date'], format='%d-%m-%Y')
df['Discharge_Date'] = pd.to_datetime(df['Discharge_Date'], format='%d-%m-%Y')
```

```python
df['Hospital_Stay_Days'] = (df['Discharge_Date'] - df['Admission_Date']
).dt.days
print(df[['Admission_Date', 'Discharge_Date', 'Hospital_Stay_Days']].head())
```

```python
print("Billing Amount:")
print(df['Billing_Amount'].describe())
```

```python
print("Hospital Stay Days:")
print(df['Hospital_Stay_Days'].describe())
```

```python
print("\nAverage Hospital Stay:")
print(df['Hospital_Stay_Days'].mean())
```
